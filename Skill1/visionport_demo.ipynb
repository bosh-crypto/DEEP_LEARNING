```python
from getpass import getpass
from pyngrok import ngrok
import threading
import uvicorn
import visionport

# Add CORS support so the HTML frontend can access Colab
from fastapi.middleware.cors import CORSMiddleware

visionport.app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"]
)

# Connect ngrok
token = getpass("Enter your ngrok authentication token: ")
ngrok.set_auth_token(token)

public_url = ngrok.connect(8000).public_url

# Start FastAPI
config = uvicorn.Config(
    visionport.app,
    host="0.0.0.0",
    port=8000
)

server = uvicorn.Server(config)

thread = threading.Thread(
    target=server.run,
    daemon=True
)
thread.start()

print("VisionPort backend is running!")
print("Backend URL:", public_url)
print("API documentation:", public_url + "/docs")
```

```python
from google.colab import files

html = open("visionport.html", "r").read()

html = html.replace(
    "http://localhost:8000/detect/",
    public_url + "/detect/"
)

with open("visionport_colab.html", "w") as f:
    f.write(html)

print("HTML connected to Colab backend!")
files.download("visionport_colab.html")
```
